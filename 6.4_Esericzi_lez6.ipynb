1. Quanti videogiochi sono stati pubblicati
2. Quali sono i generi più popolari? Mostralo con un bar plot
3. Plotta l'evoluzione del numero di giochi pubblicati nel tempo
4. Prova a riprodurre questo plot 




```python
# Import dei pacchetti
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Caricamento del dataset
url = 'https://zenodo.org/record/5898311/files/vgsales.csv'
df = pd.read_csv(url)
df_cleaned = df.drop_duplicates()
df.head()
```

```python
published_vg = df.shape[0]
published_vg
```

```python
genre_count = df_cleaned["Genre"].value_counts()
sns.set_style("darkgrid")
plt.xticks(rotation=90)
sns.barplot(data=genre_count, color="lightblue", )
```

```python

#count_by_year = df_cleaned["Year"].value_counts().sort_index()

sns.histplot(data=df_cleaned,x="Year", kde=True, discrete=True)
```

```python
sales_by_genre = df_cleaned.groupby(by="Genre")[["JP_Sales", "NA_Sales", "EU_Sales","Other_Sales","Global_Sales"]].sum().sort_values("Global_Sales", ascending=False).head(4)

sales_by_genre[["NA_Sales", "JP_Sales", "EU_Sales","Other_Sales"]].plot(kind="bar", stacked=True, color=["red", "blue", "green","orange"])
```
