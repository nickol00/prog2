### **1. Trasformare cicli in list comprehension**

Fai tre esempi di cicli producono una lista e poi riscrivi lo stesso codice usando la list comprehension

```python
li1 = list()
li2 = list()
li3 = list()

for i in range(1,20,3):
    li1.append(i)
for i in range(1,20):
    if i%2==0:
        li2.append(i)
for i in range(1,len(li2)):
    if(li2[i] in li1):
        li3.append(li2[i])
print(li1)
print(li2)
print(li3)
print("-------------")
com1 = [ i for i in range(1,20,3)]
com2 = [ i for i in range(1,20) if i%2==0]
com3 = [ i for i in com1 if i in com2]
print(com1)
print(com2)
print(com3)
```

3. Quale pensi sia il tipo di dato (`dtype`) del vettore?  Prova a rispondere senza eseguire il codice e Verifica la tua risposta accedendo all'attributo `.dtype` del vettore

4. Scrivi l'array usando una list comprehension che controlla che i numeri siano primi. 

```python
import numpy as np

primi_arr = np.array([2, 3, 5, 7])

print(len(primi_arr))
print(primi_arr.size)

print(primi_arr.dtype)

def is_prime(n:int):
    if n==1: return False
    for i in range(2,int(n**0.5 +1)):
        if n%i==0: return False
    return True
primevec = [ i for i in range(1,10) if is_prime(i)]
print(primevec)
```
